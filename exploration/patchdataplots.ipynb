```python
def load_data(patch_idx):
    # load the .h5 data file into a dataframe, uses patch index (0-20)
    df = pd.read_hdf(f"../gaia-data/cleaned-data/gd1_patch{patch_idx}.h5")
    print(f"Data for Patch {patch_idx}")
    return df


def plot_data(df):
    
    # create figure with 1 row and 3 columns for data subplots
    fig, axes = plt.subplots(1, 3, figsize=(15, 5), tight_layout=True)
    fig.suptitle('Full Patch Data')
    color = 'Greys'

    # plot position coordinates
    binspos = np.linspace(-15,15,100), np.linspace(-15,15,100)
    h = axes[0].hist2d(df['rotra'], df['rotdec'], bins=binspos, cmap=color, cmin=0, vmax=250)
    axes[0].set_xlabel('Rotated Right Ascension ϕ', fontsize = 10)
    axes[0].set_ylabel('Rotated Declination λ', fontsize = 10)
    c = fig.colorbar(h[3], ax=axes[0])
    c.ax.set_title('Counts', fontsize=8)


    # plot proper motion coordinates
    binspm = (np.linspace(-20,20,100), np.linspace(-20,20,100))
    h = axes[1].hist2d(df['rotpmra'], df['rotpmdec'], bins=binspm, cmap=color, cmin=0)
    axes[1].set_xlabel('Rotated Right Ascension Proper Motion μ_ϕcosλ', fontsize = 10)
    axes[1].set_ylabel('Rotated Declination Proper Motion μ_λ', fontsize = 10)  
    c = fig.colorbar(h[3], ax=axes[1])
    c.ax.set_title('Counts', fontsize=8)

    # plot photometric figures
    binsfeat = (np.linspace(0,3,100), np.linspace(10,21,100))
    h = axes[2].hist2d(df['b-r'], df['g'], bins=binsfeat, cmap=color, cmin=0)
    axes[2].set_xlabel('Color b-r', fontsize = 10)
    axes[2].set_ylabel('Magnitude g', fontsize = 10)
    c = fig.colorbar(h[3], ax=axes[2])
    c.ax.set_title('Counts', fontsize=8)

    plt.show()

# for idx in range(0, 21):
#     df = load_data(idx)
#     plot_data(df)
```

```python
df_stars = df_patch0[df_patch0['stream'] == True]
df_background = df_patch0[df_patch0['stream'] == False]

len_df = len(df_patch0)
len_stars = len(df_stars)
len_background = len(df_background)
total = len_stars + len_background

print(len_df, len_stars, len_background, total)
```

```python
# rotated proper motion (right ascension) -- mu phi cos lambda
mu_phi_cos_lam = df_patch0['rotpmra']

plt.hist(mu_phi_cos_lam, bins=100, range=(-40,40))
plt.xlabel('Rotated and Centered Proper Motion (RA) mu_phi*')
plt.ylabel('Number of Stars')
plt.title('All Stars in Patch 0')
plt.show()

plt.hist(df_stars['rotpmra'], bins=100, range=(-40,40))
plt.xlabel('Rotated and Centered Proper Motion (RA) mu_phi*')
plt.ylabel('Number of Stars')
plt.title('Stream Stars in Patch 0')
plt.show()

plt.hist(df_background['rotpmra'], bins=100, range=(-40,40))
plt.xlabel('Rotated and Centered Proper Motion (RA) mu_phi*')
plt.ylabel('Number of Stars')
plt.title('Background Stars in Patch 0')
plt.show()
```

```python
# rotated proper motion (declination) -- mu lambda ******

mu_lam = df_patch0['rotpmdec']

plt.hist(mu_lam, bins=100, range=(-50,10))
plt.xlabel('Rotated and Centered Proper Motion (DEC) mu_lambda')
plt.ylabel('Number of Stars')
plt.title('All Stars in Patch 0')
plt.show()

plt.hist(df_stars['rotpmdec'], bins=100, range=(-18,-8))
plt.xlabel('Rotated and Centered Proper Motion (DEC) mu_lambda')
plt.ylabel('Number of Stars')
plt.title('Stream Stars in Patch 0')
plt.show()

plt.hist(df_background['rotpmdec'], bins=100, range=(-17,-10))
plt.xlabel('Rotated and Centered Proper Motion (DEC) mu_lambda')
plt.ylabel('Number of Stars')
plt.title('Background Stars in Patch 0')
plt.show()
```

```python
### try dividing into signal/sideband regions, -3:3 std sideband, -1:1 signal
### using mu_lambda (mu_lam)

signal_parameter = df_stars['rotpmdec']

median = np.median(signal_parameter)
std = np.std(signal_parameter)
print('Signal Parameter mu_lam median = ', median)
print('Signal Parameter mu_lam standard deviation = ', std)

sig_factor = 1
sb_factor = 3

sig_low = median - sig_factor*std
sig_high = median + sig_factor*std
sb_low = median - sb_factor*std
sb_high = median + sb_factor*std

print('Signal Region Range: ', sig_low, ' to ', sig_high)
print('Sideband Region Range: ', sb_low, ' to ', sig_low, ' and ', sig_high, ' to ', sb_high)
```

```python
stream_sig_region = df_stars[(df_stars['rotpmdec'] >= sig_low) & (df_stars['rotpmdec'] <= sig_high)]
stream_sb_region = df_stars[((df_stars['rotpmdec'] >= sb_low) & (df_stars['rotpmdec'] < sig_low)) | ((df_stars['rotpmdec'] <= sb_high) & (df_stars['rotpmdec'] > sig_high))]

background_sig_region = df_background[(df_background['rotpmdec'] >= sig_low) & (df_background['rotpmdec'] <= sig_high)]
background_sb_region = df_background[((df_background['rotpmdec'] >= sb_low) & (df_background['rotpmdec'] < sig_low)) | ((df_background['rotpmdec'] <= sb_high) & (df_background['rotpmdec'] > sig_high))]

outer_region = df_patch0[(df_patch0['rotpmdec'] < sb_low) | (df_patch0['rotpmdec'] > sb_high)]

print(f'Signal Region has {len(stream_sig_region) + len(background_sig_region)} stars, {len(stream_sig_region)} stream and {len(background_sig_region)} background.')
print(f'Sideband Region has {len(stream_sb_region) + len(background_sb_region)} stars, {len(stream_sb_region)} stream and {len(background_sb_region)} background.')      
print(f'Outer Region has {len(outer_region)} stars.')
```

```python
## fix bins

bins = np.linspace(sb_low - (sig_low - sb_low), sb_high + (sb_high - sig_high), 50)

plt.hist(stream_sig_region['rotpmdec'], label='Signal Region (GD-1)', color='red', alpha=1, edgecolor='white', bins=bins)
plt.hist(stream_sb_region['rotpmdec'], label='Sideband Region (GD-1)', color='red', alpha=0.5, edgecolor='white', bins=bins)
plt.xlabel('Rotated and Centered Proper Motion (DEC) mu_lambda')
plt.ylabel('Number of Stars')
plt.title('Stream Stars in Patch 0')
plt.legend()
plt.show()

plt.hist(background_sig_region['rotpmdec'], label='Signal Region (background)', color='grey', alpha=1, bins=bins)
plt.hist(background_sb_region['rotpmdec'], label='Sideband Region (background)', color='grey', alpha=0.5, bins=bins)
plt.xlabel('Rotated and Centered Proper Motion (DEC) mu_lambda')
plt.ylabel('Number of Stars')
plt.title('Background Stars in Patch 0')
plt.legend()
plt.show()
```
