```python
import pandas as pd
```

```python
df_test_3 = pd.read_hdf('../results/patch_3/rotpmdec/training/df_test.h5')
print(len(df_test_3))
```

```python
df_test_4 = pd.read_hdf('../results/patch_4/rotpmdec/training/df_test.h5')
print(len(df_test_4))
```

```python
df_test_check = pd.concat([df_test_3, df_test_4], ignore_index=True)
print(len(df_test_check))
```

```python
check_cols = ['b-r', 'g', 'ra', 'dec', 'pmra']
duplicate_rows = df_test_check[df_test_check.duplicated(subset=check_cols, keep=False)].sort_values(by=check_cols)
print(f'Duplicate rows found: {len(duplicate_rows)}')
print(duplicate_rows)
```

```python
df_test_check = df_test_check.drop_duplicates(subset=check_cols,ignore_index=True)
print(f'Number of unique rows after dropping duplicates: {len(df_test_check)}')
```
