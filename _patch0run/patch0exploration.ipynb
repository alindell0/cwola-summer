```python
def fiducial_selections(df): 
    # CWOLA Stellar Stream fiducial cuts https://arxiv.org/pdf/2305.03761
    df = df[df['g'] < 20.2] # ensures uniform acceptance by Gaia satellite
    df = df[(np.abs(df['rotpmdec']) > 2) | (np.abs(df['rotpmra']) > 2)] # remove distant stars concentrated near zero proper motion, not equally distributed throughout patch
    df = df[(df['b-r'] >= 0.5) & (df['b-r'] <=1 )] # isolates old and low-metallicity stellar streams in color space

    return df
```

```python
df_updated = fiducial_selections(df_test)
len(df_updated)
```

```python
df_ranked = df_updated.sort_values(by='nn_score', ascending=False)
df_top_250 = df_ranked[:250]
df_top_250.head()
```

```python
purity = len(df_top_250[df_top_250['stream']==True]) / len(df_top_250)
print(f'{purity * 100} %')
```

```python
completeness = len(df_top_250[df_top_250['stream']==True])/len(df_test[df_test['stream']==True])
print(f'{completeness * 100} %')
```
