```python
import matplotlib.pyplot as plt

import sys
sys.path.append('../scripts')
from functions import *
```

```python
## checking distribution of signal and sideband regions for both stream and background stars
## confirm indistinguishability based on choice of regions
```

```python
patch_0 = load_data(0)
```

```python
regions_dec = signal_sideband(patch_0, save_folder = None, pm_parameter='rotpmdec', sig_factor=1, sb_factor=3, bin_num=51)
```

```python
# plot signal/sideband stream/background for each input feature

df_signalreg = regions_dec[regions_dec['region_label']==1]
df_signalreg_stream = df_signalreg[df_signalreg['stream']==True]
df_signalreg_back = df_signalreg[df_signalreg['stream']==False]

df_backgroundreg = regions_dec[regions_dec['region_label']==0]
df_backgroundreg_stream = df_backgroundreg[df_backgroundreg['stream']==True]
df_backgroundreg_back = df_backgroundreg[df_backgroundreg['stream']==False]

for var in ['rotra', 'rotdec', 'rotpmra', 'b-r', 'g']:
    plt.hist(df_signalreg_stream[var], label='Signal Region (GD-1)', histtype='step', density=True, color='red')
    plt.hist(df_backgroundreg_stream[var], label='Sideband Region (GD-1)', histtype='step', density=True, color='red', alpha=0.5)
    plt.hist(df_signalreg_back[var], label='Signal Region (Background)', histtype='step', density=True, color='grey')
    plt.hist(df_backgroundreg_back[var], label='Sideband Region (Background)', histtype='step', density=True, color='grey', alpha=0.5)
    plt.xlabel(f'{var}')
    plt.ylabel('Normalized Counts')
    plt.legend()
    plt.show()
```
