### General numbers for 21 patches

```python
import pandas as pd
```

```python
df = pd.read_hdf("../gaia-data/cleaned-data/gd1_patch0.h5")
df.head()
```

```python
total_stream_stars_per_patch = 0

for idx in range(21):
    df = pd.read_hdf(f"../gaia-data/cleaned-data/gd1_patch{idx}.h5")
    print(f"Number of stars in patch {idx}:")
    print(len(df))
    
    stream_stars = 0
    for value in df['stream']:
        if value == True:
            stream_stars += 1
            total_stream_stars_per_patch += 1
        else:
            continue
    print("Number of GD-1 identified stars:")
    print(stream_stars)
    print("----------------------")

avg_stream_stars_per_patch = total_stream_stars_per_patch / 21

print("Average number of GD-1 stars per patch =")
print(avg_stream_stars_per_patch)
```
